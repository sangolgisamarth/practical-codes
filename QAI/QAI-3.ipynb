```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
```

```python
# Create 3 qubits and 3 classical bits
qc = QuantumCircuit(3, 3)
```

```python
# Step 1: Prepare the unknown state on qubit 0
qc.h(0)
```

```python
# Step 2: Create entanglement between qubit 1 and qubit 2
qc.h(1)
qc.cx(1, 2)
```

```python
# Step 3: Alice performs Bell measurement operations
qc.cx(0, 1)
qc.h(0)
```

```python
# Step 4: Measure Alice's qubits
qc.measure(0, 0)
qc.measure(1, 1)
```

```python
# Step 5: Apply correction gates to Bob's qubit
# Conditional X correction
with qc.if_test((qc.clbits[1], 1)):
    qc.x(2)
```

```python
# Conditional Z correction
with qc.if_test((qc.clbits[0], 1)):
    qc.z(2)
```

```python
# Step 6: Measure Bob's qubit
qc.measure(2, 2)
```

```python
# Display circuit
print(qc.draw())
```

```python
# Create simulator
simulator = AerSimulator()

# Transpile circuit
compiled_circuit = transpile(qc, simulator)

# Execute circuit
result = simulator.run(
    compiled_circuit,
    shots=1000
).result()
```

```python
# Retrieve results
counts = result.get_counts()

print("Teleportation Measurement Results:")
print(counts)
```
