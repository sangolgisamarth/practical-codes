```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
```

```python
# Create a 16-qubit quantum circuit
qc = QuantumCircuit(16, 16)
```

```python
# Apply Hadamard gate to all 16 qubits
for i in range(16):
    qc.h(i)
```

```python
# Measure all qubits
qc.measure(range(16), range(16))
```

```python
# Display circuit
print(qc.draw())
```

```python
# Create quantum simulator
simulator = AerSimulator()
```

```python
# Execute circuit
compiled_circuit = transpile(qc, simulator)
```

```python
result = simulator.run(
    compiled_circuit,
    shots=1
).result()
```

```python
# Get random binary number
counts = result.get_counts()
```

```python
random_binary = list(counts.keys())[0]
```

```python
# Convert binary to decimal
random_decimal = int(random_binary, 2)
print("Random 16-bit Binary Number:", random_binary)
print("Random Decimal Number:", random_decimal)
```
