```python
from qiskit import QuantumCircuit, transpile
from qiskit.circuit.library import QFT
from qiskit_aer import AerSimulator
```

```python
# Create a 5-qubit quantum circuit with 5 classical bits
qc = QuantumCircuit(5, 5)
```

```python
# Prepare the initial quantum state
qc.h(0)
qc.x(1)
```

```python
# Apply 5-qubit Quantum Fourier Transform
qft = QFT(5)
qc.compose(qft, inplace=True)
```

```python
# Measure all qubits
qc.measure(range(5), range(5))
```

```python
# Create quantum simulator
sim = AerSimulator()
```

```python
# Transpile and execute the circuit
compiled_circuit = transpile(qc, sim)

result = sim.run(compiled_circuit, shots=1024).result()
```

```python
# Display measurement results
print("QFT Results:", result.get_counts())
```

```python
# Display quantum circuit
print(qc.draw())
```
