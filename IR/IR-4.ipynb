```python
import networkx as nx
```

```python
# Create a directed graph
G = nx.DiGraph()
```

```python
# Add web page links
G.add_edges_from([
    ('A', 'B'),
    ('A', 'C'),
    ('B', 'C'),
    ('C', 'A'),
    ('D', 'C')
])
```

```python
# Calculate PageRank
pagerank = nx.pagerank(G, alpha=0.85)
```

```python
# Display PageRank
print("PageRank of Web Pages:")

for page, rank in pagerank.items():
    print(page, ":", round(rank, 4))
```

```python
# Display ranking in descending order
print("\nWeb Page Ranking:")

for page, rank in sorted(pagerank.items(), key=lambda x: x[1], reverse=True):
    print(page, ":", round(rank, 4))
```
