```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize
import warnings
warnings.filterwarnings('ignore')
```

```python
text = """Information retrieval is the process of obtaining relevant
documents from a large
collection. Search engines are retrieving and ranking documents according
to the queries
entered by the users. Pre-processing of text is an important step in
retrieval systems!"""
print("Original text:\n", text)
```

```python
# 1. Lowercase and remove punctutation / numbers
clean = re.sub(r"[^a-z\s]","",text.lower())
print("After cleaning:",clean)
```

```python
# 2. Tokenization 
tokens = word_tokenize(clean)
print("Tokens:",tokens)
```

```python
# 3. stop word removal
stop_words=(stopwords.words("english"))
filtered=[w for w in tokens if w not in stop_words]
print("\nAfter stop word removal:\n", filtered)
```

```python
# 4. stemming
ps=PorterStemmer()
stemmed= [ps.stem(w) for w in filtered]
print("\nAfter stemming:\n", stemmed)
```

```python
print("word -> stem")
for w,s in zip(filtered,stemmed):
    print(w,"->",s)
```
