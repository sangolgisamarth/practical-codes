```python

# Initialize inverted index
index = {}

# Function to add documents
def add_document(doc_id, text):
    tokens = text.lower().split()

    for token in set(tokens):
        if token not in index:
            index[token] = []

        index[token].append(doc_id)


# Function to search documents
def search(query):
    tokens = query.lower().split()

    if not tokens:
        return []

    result_docs = set(index.get(tokens[0], []))

    for token in tokens[1:]:
        result_docs = result_docs.intersection(
            set(index.get(token, []))
        )

    return sorted(result_docs)


# Main Program
add_document(1, "information retrieval is fun")
add_document(2, "retrieval of documents using inverted files")
add_document(3, "information systems and data retrieval")

# Display inverted index
print("Inverted Index:")
for word, docs in index.items():
    print(word, ":", docs)

# Search documents
print("\nSearch 'retrieval':", search("retrieval"))

print("Search 'information retrieval systems':",
      search("information retrieval systems"))
```
