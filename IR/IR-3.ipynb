```python
import pandas as pd
import numpy as np

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.parameter_estimator import DiscreteMLE
from pgmpy.inference import VariableElimination
```

```python
np.random.seed(42)
data = pd.DataFrame({
    'Age': np.random.choice(['Young', 'Middle', 'Old'], size=500),
    'Cholesterol': np.random.choice(['Normal', 'High'], size=500),
    'HeartDisease': np.random.choice(['Yes', 'No'], size=500, p=[0.4, 0.6])
})
```

```python
data
```

```python
# Create network
model = DiscreteBayesianNetwork([
    ('Age', 'HeartDisease'),
    ('Cholesterol', 'HeartDisease')
])

# Train model
model.fit(data, estimator=DiscreteMLE())

# Validate model
print("Model Check Validated:", model.check_model())
```

```python
# Inference
inference = VariableElimination(model)

result = inference.query(
    variables=['HeartDisease'],
    evidence={
        'Age': 'Old',
        'Cholesterol': 'High'
    }
)

print("\nDiagnosis probability distribution:")
print(result)
```
